```python
import pandas as pd
```

```python
df = pd.read_csv("../gender-pay-gap-dataset-final.csv")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming 'df' is your DataFrame and it contains 'year', 'sex', and 'salary' columns

# Group by year and sex, then calculate average salary
avg_salary_by_year_sex = df.groupby(['year', 'sex'])['inflated_wage'].mean().reset_index()

# Pivot the data to get years as index and sex as columns
salary_pivot = avg_salary_by_year_sex.pivot(index='year', columns='sex', values='inflated_wage')

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(salary_pivot.index, salary_pivot['Male'], label='Male', marker='o')
plt.plot(salary_pivot.index, salary_pivot['Female'], label='Female', marker='o')
plt.title('Average Salary by Year and Gender')
plt.xlabel('Year')
plt.ylabel('Average Salary')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
